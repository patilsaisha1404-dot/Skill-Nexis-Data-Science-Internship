### Question 1: Load a CSV file using Pandas and display basic info.

```python
import pandas as pd

# Load the dataset
df = pd.read_csv('SQL_Sales_Dataset_200_Rows.xlsx - Sheet1.csv')

# Display basic info
df.info()
df.head()
```

### Question 2: Handle missing values and duplicates using Pandas.

```python
print("Missing values:", df.isnull().sum())
print("Duplicates:", df.duplicated().sum())

df = df.drop_duplicates()
df = df.fillna(method='ffill')
```

### Question 3: Group data by category and find total revenue.

```python
total_revenue = df.groupby('category')['total_price'].sum().reset_index()
total_revenue
```

### Question 4: Sort data by multiple columns using Python.

```python
sorted_df = df.sort_values(by=['category', 'total_price'], ascending=[True, False])
sorted_df.head()
```

### Question 5: Create a correlation matrix for numerical columns.

```python
numeric_df = df.select_dtypes(include=['number'])
correlation_matrix = numeric_df.corr()
correlation_matrix
```
